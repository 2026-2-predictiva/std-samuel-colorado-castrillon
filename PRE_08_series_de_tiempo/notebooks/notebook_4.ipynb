```python
# Model 4: linear autoregressive model, AR(13), one-step-ahead forecasts
from sklearn.linear_model import LinearRegression

P = 13
df = make_lagged_ts(df, P, "yt_true")
x_columns = [f"lagged_{i}m" for i in range(1, P + 1)]

data = df.dropna()
X_complete, y_complete, X_train, y_train, X_test, y_test = train_test_split(
    data, x_columns, "yt_true"
)

model = LinearRegression().fit(X_train, y_train)
df["yt_pred_ar"] = np.nan
df.loc[data.index, "yt_pred_ar"] = model.predict(X_complete)
plot_time_series(df[["yt_true", "yt_pred_ar"]])
```
